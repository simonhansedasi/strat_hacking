## Format

Each dog is a dict with three fields:

| Field | Type | Description |
|-------|------|-------------|
| `name` | string | Dog's name/breed printed on the card |
| `spaces` | list of int | Die values shown as red squares (1–6), one per space |
| `paw_prints` | int | Number of 🐾 symbols on the card |

**Example:** a dog with spaces showing 1, 3, 3, 5 and one paw print:
```python
{"name": "Beagle", "spaces": [1, 3, 3, 5], "paw_prints": 1}
```

Duplicate values in `spaces` are fine — they mean the card has two separate spaces that each require that value.

## Preview & validate

Run this to check your entries look right before exporting.

````python
errors = []
for i, dog in enumerate(DOGS):
    if not dog.get("name"):
        errors.append(f"Dog {i}: missing name")
    spaces = dog.get("spaces", [])
    if not spaces:
        errors.append(f"Dog {i} ({dog.get('name')}): no spaces defined")
    if any(s not in range(1, 7) for s in spaces):
        errors.append(f"Dog {i} ({dog.get('name')}): spaces must be 1–6, got {spaces}")
    if not isinstance(dog.get("paw_prints", None), int):
        errors.append(f"Dog {i} ({dog.get('name')}): paw_prints must be an integer")

if errors:
    print("⚠️  Fix these before exporting:")
    for e in errors:
        print(f"  • {e}")
else:
    print(f"✓  {len(DOGS)} dogs look good\n")
    # Summary table
    print(f"{'Name':<22} {'Spaces':<28} Paw prints  Total spaces")
    print("-" * 70)
    for dog in DOGS:
        spaces_str = str(sorted(dog['spaces']))
        print(f"{dog['name']:<22} {spaces_str:<28} {dog['paw_prints']:<12} {len(dog['spaces'])}")
````

## Export to JSON

Writes `../data/dogs.json`. The simulation will load from there.

````python
import json, os

if errors:
    print("⚠️  Fix validation errors above before exporting.")
else:
    out_path = os.path.join(os.path.dirname(os.getcwd()), "data", "dogs.json")
    os.makedirs(os.path.dirname(out_path), exist_ok=True)
    with open(out_path, "w") as f:
        json.dump(DOGS, f, indent=2)
    print(f"✓  Exported {len(DOGS)} dogs → {out_path}")
````
